## Notebook 03 - Feature Engineering

### Objective
This notebook creates meaningful features from historical silver price data to improve the performance of forecasting and machine learning models. Technical indicators, lag variables, rolling statistics, and time-based features are generated.

```python
# =====================================================
# Import Required Libraries
# =====================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# =====================================================
# Load Cleaned Dataset
# =====================================================

silver_daily = pd.read_csv(
    "../outputs/clean_silver_daily.csv",
    parse_dates=["Date"]
)

print("Dataset Loaded Successfully")
print(silver_daily.shape)
```

```python
# =====================================================
# Dataset Preview
# =====================================================

silver_daily.head()
```

```python
# =====================================================
# Time Features
# =====================================================

silver_daily["Year"] = silver_daily["Date"].dt.year
silver_daily["Month"] = silver_daily["Date"].dt.month
silver_daily["Quarter"] = silver_daily["Date"].dt.quarter
silver_daily["Day"] = silver_daily["Date"].dt.day
silver_daily["DayOfWeek"] = silver_daily["Date"].dt.dayofweek
silver_daily["WeekOfYear"] = silver_daily["Date"].dt.isocalendar().week.astype(int)

print("Time features created successfully.")
```

```python
# =====================================================
# Lag Features
# =====================================================

silver_daily["Lag_1"] = silver_daily["Close"].shift(1)
silver_daily["Lag_3"] = silver_daily["Close"].shift(3)
silver_daily["Lag_7"] = silver_daily["Close"].shift(7)
silver_daily["Lag_14"] = silver_daily["Close"].shift(14)
silver_daily["Lag_30"] = silver_daily["Close"].shift(30)

print("Lag features created successfully.")
```

```python
# =====================================================
# Rolling Mean
# =====================================================

silver_daily["MA_7"] = silver_daily["Close"].rolling(7).mean()
silver_daily["MA_14"] = silver_daily["Close"].rolling(14).mean()
silver_daily["MA_30"] = silver_daily["Close"].rolling(30).mean()

print("Moving averages created successfully.")
```

```python
# =====================================================
# Rolling Volatility
# =====================================================

silver_daily["Volatility_7"] = silver_daily["Close"].rolling(7).std()
silver_daily["Volatility_14"] = silver_daily["Close"].rolling(14).std()
silver_daily["Volatility_30"] = silver_daily["Close"].rolling(30).std()

print("Rolling volatility calculated.")
```

```python
# =====================================================
# Price Change Features
# =====================================================

silver_daily["Daily_Change"] = silver_daily["Close"] - silver_daily["Open"]

silver_daily["High_Low_Range"] = (
    silver_daily["High"] - silver_daily["Low"]
)

silver_daily["Open_Close_Range"] = (
    silver_daily["Open"] - silver_daily["Close"]
)

print("Price change features created.")
```

```python
# =====================================================
# Exponential Moving Average
# =====================================================

silver_daily["EMA_7"] = silver_daily["Close"].ewm(span=7, adjust=False).mean()

silver_daily["EMA_14"] = silver_daily["Close"].ewm(span=14, adjust=False).mean()

silver_daily["EMA_30"] = silver_daily["Close"].ewm(span=30, adjust=False).mean()

print("EMA features created successfully.")
```

```python
# =====================================================
# Relative Strength Index (RSI)
# =====================================================

delta = silver_daily["Close"].diff()

gain = delta.clip(lower=0)

loss = -delta.clip(upper=0)

avg_gain = gain.rolling(14).mean()

avg_loss = loss.rolling(14).mean()

rs = avg_gain / avg_loss

silver_daily["RSI_14"] = 100 - (100 / (1 + rs))

print("RSI calculated successfully.")
```

```python
# =====================================================
# MACD
# =====================================================

ema12 = silver_daily["Close"].ewm(span=12, adjust=False).mean()

ema26 = silver_daily["Close"].ewm(span=26, adjust=False).mean()

silver_daily["MACD"] = ema12 - ema26

silver_daily["MACD_Signal"] = silver_daily["MACD"].ewm(span=9, adjust=False).mean()

print("MACD created successfully.")
```

```python
# =====================================================
# Bollinger Bands
# =====================================================

rolling_mean = silver_daily["Close"].rolling(20).mean()

rolling_std = silver_daily["Close"].rolling(20).std()

silver_daily["BB_Middle"] = rolling_mean

silver_daily["BB_Upper"] = rolling_mean + (2 * rolling_std)

silver_daily["BB_Lower"] = rolling_mean - (2 * rolling_std)

print("Bollinger Bands created successfully.")
```

```python
# =====================================================
# Remove Missing Values
# =====================================================

silver_daily = silver_daily.dropna().reset_index(drop=True)

print("Final Dataset Shape :", silver_daily.shape)
```
