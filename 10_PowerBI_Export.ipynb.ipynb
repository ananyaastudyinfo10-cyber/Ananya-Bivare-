```python
# =====================================================
# Import Libraries
# =====================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

print("Libraries Imported Successfully")
```

```python
# =====================================================
# Load Files
# =====================================================

silver = pd.read_csv("../outputs/feature_engineered_silver.csv")

comparison = pd.read_csv("../outputs/model_comparison.csv")

risk = pd.read_csv("../outputs/risk_summary.csv")

ensemble = pd.read_csv("../outputs/ensemble_predictions.csv")

print("All files loaded successfully.")
```

```python
# =====================================================
# Dashboard Summary
# =====================================================

summary = pd.DataFrame({

"Total Records":[len(silver)],

"Models Compared":[len(comparison)],

"Average Closing Price":[silver["Close"].mean()],

"Highest Closing Price":[silver["Close"].max()],

"Lowest Closing Price":[silver["Close"].min()],

"Average Volume":[silver["Volume"].mean()]

})

summary
```

```python
# =====================================================
# Monthly Average Price
# =====================================================

silver["Date"] = pd.to_datetime(silver["Date"])

silver["Year"] = silver["Date"].dt.year

silver["Month"] = silver["Date"].dt.month

monthly = silver.groupby(

["Year","Month"]

)["Close"].mean().reset_index()

monthly.head()
```

```python
# =====================================================
# Yearly Average
# =====================================================

yearly = silver.groupby(

"Year"

)["Close"].mean().reset_index()

yearly.head()
```

```python
# =====================================================
# Volume Summary
# =====================================================

volume = silver.groupby(

"Year"

)["Volume"].sum().reset_index()

volume.head()
```

```python
# =====================================================
# Technical Indicators
# =====================================================

technical = silver[

[

"Date",

"Close",

"MA_7",

"MA_14",

"MA_30",

"EMA_7",

"EMA_14",

"EMA_30",

"RSI_14",

"MACD"

]

]

technical.head()
```

```python
# =====================================================
# Export Files
# =====================================================

summary.to_csv(

"../outputs/dashboard_summary.csv",

index=False

)

monthly.to_csv(

"../outputs/monthly_price.csv",

index=False

)

yearly.to_csv(

"../outputs/yearly_price.csv",

index=False

)

volume.to_csv(

"../outputs/yearly_volume.csv",

index=False

)

technical.to_csv(

"../outputs/technical_indicators.csv",

index=False

)

print("Dashboard datasets exported successfully.")
```
