## Notebook 01 - Data Loading & Dataset Inspection

**Objective:**
This notebook loads all datasets, performs an initial inspection, checks data quality, and summarizes the datasets before further analysis.

```python
# ==========================================
# Import Required Libraries
# ==========================================

import os
import warnings

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
# ==========================================
# Display Current Working Directory
# ==========================================

print("Current Working Directory:")
print(os.getcwd())
```

```python
# ==========================================
# Load All Project Datasets
# ==========================================

# Dataset folder path
data_path = "../datasets/"

# Load datasets
silver_daily = pd.read_csv(data_path + "silver_daily_ohlcv_2000_2025.csv")

macroeconomic = pd.read_csv(data_path + "silver_macroeconomic_monthly.csv")

futures = pd.read_csv(data_path + "silver_futures_contracts.csv")

supply_demand = pd.read_csv(data_path + "silver_supply_demand_annual.csv")

sentiment = pd.read_csv(data_path + "silver_sentiment_weekly.csv")

print("✅ All datasets loaded successfully.")
```

```python
# Dataset Dimensions

#The following output displays the number of rows and columns in each dataset. This helps verify that the files have been loaded correctly.
# ==========================================
# Dataset Shapes
# ==========================================

print("Silver Daily Dataset      :", silver_daily.shape)
print("Macroeconomic Dataset     :", macroeconomic.shape)
print("Futures Dataset           :", futures.shape)
print("Supply-Demand Dataset     :", supply_demand.shape)
print("Sentiment Dataset         :", sentiment.shape)
```

```python
# ==========================================
# Display First Five Rows
# ==========================================

datasets = {
    "Silver Daily": silver_daily,
    "Macroeconomic": macroeconomic,
    "Futures": futures,
    "Supply Demand": supply_demand,
    "Sentiment": sentiment
}

for name, df in datasets.items():
    print("=" * 80)
    print(f"{name} Dataset")
    print("=" * 80)
    display(df.head())
```

```python
# ==========================================
# Dataset Information
# ==========================================

for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(f"{name} Dataset Information")
    print("=" * 80)
    df.info()
```

```python
# ==========================================
# Missing Values
# ==========================================

for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(f"Missing Values in {name} Dataset")
    print("=" * 80)
    
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) == 0:
        print("No missing values found.")
    else:
        display(missing)
```

```python
# ==========================================
# Dataset Summary Table
# ==========================================

summary = pd.DataFrame({
    "Dataset": [
        "Silver Daily",
        "Macroeconomic",
        "Futures",
        "Supply Demand",
        "Sentiment"
    ],
    "Rows": [
        silver_daily.shape[0],
        macroeconomic.shape[0],
        futures.shape[0],
        supply_demand.shape[0],
        sentiment.shape[0]
    ],
    "Columns": [
        silver_daily.shape[1],
        macroeconomic.shape[1],
        futures.shape[1],
        supply_demand.shape[1],
        sentiment.shape[1]
    ],
    "Missing Values": [
        silver_daily.isnull().sum().sum(),
        macroeconomic.isnull().sum().sum(),
        futures.isnull().sum().sum(),
        supply_demand.isnull().sum().sum(),
        sentiment.isnull().sum().sum()
    ]
})

summary
```

```python
# ==========================================
# Save Summary
# ==========================================

import os

os.makedirs("../outputs", exist_ok=True)

summary.to_csv("../outputs/dataset_summary.csv", index=False)

print("✅ Dataset summary saved successfully!")
```
